### Step 3. Assign it to a variable called chipo.

```python
#importing required libraries
import os
import time
import shutil

import pandas as pd

file='chipotle.csv'
srcPath='/home/dreams/Projects/webScraping/MachineLearningProject/pandasPractices/pandas_exercises/01_Getting_&_Knowing_Your_Data/Chipotle'
#copying current csv file
if not os.path.exists(os.path.join(os.getcwd(),file)):
    #copying current chipotle csv
    # shutil.copy(srcPath,os.getcwd())
    shutil.copy(os.path.join(srcPath,file),os.getcwd())
    time.sleep(1)
    print('copied to  current path')
else:
    #reading current dataFrame
    chipo=pd.read_csv(file)
# chipo.head(10)
chipo['choice_description'].values[4]




```

### Step 4. How many products cost more than $10.00?

```python
import numpy as np

chipo['prices_float']=chipo['item_price'].astype(str).str.replace("$","",regex=False)
chipo['prices_float']=chipo['prices_float'].astype(np.float64)

mask=(chipo['prices_float']>10)
print(len(chipo[mask]['item_name'].unique()))



```

```python
#creating new dataFrame
df=pd.DataFrame({
        'item_name':chipo['item_name'],
        'choice_description':chipo['choice_description'],
        'item_price':chipo['item_price']
    }
)
df.head()
```

### Step 6. Sort by the name of the item

```python
df.sort_values(by=['item_name'])
```
