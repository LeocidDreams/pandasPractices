### Step 3. Assign it to a variable called drinks.

```python
# !curl -O 'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv'
import pandas as pd
# drinks=pd.read_csv('drinks.csv',sep=",")
drinks=pd.read_csv('drinks.csv',sep=',',keep_default_na=False)
drinks
```

### Step 4. Which continent drinks more beer on average?

```python
# grouped=drinks.groupby(['continent'])['beer_servings'].sum().sort_values(ascending=False)
# # for name,subgroup in grouped:
# #     print(name)
# #     print(subgroup['beer_servings'].sum())
# #     break
    
# grouped

grouped=drinks.groupby(['continent'])
grouped['beer_servings'].sum().sort_values(ascending=False)
```

### Step 5. For each continent print the statistics for wine consumption.

```python
# grouped=drinks.groupby(['continent'])['wine_servings'].describe()
# grouped
grouped=drinks.groupby(['continent'])
grouped['wine_servings'].describe()
```

### Step 6. Print the mean alcohol consumption per continent for every column

```python
grouped=drinks.groupby(['continent'])

# grouped['total_litres_of_pure_alcohol'].mean()
grouped.mean(numeric_only=True)
```

### Step 7. Print the median alcohol consumption per continent for every column

```python
grouped=drinks.groupby(['continent'])
grouped['total_litres_of_pure_alcohol'].median()
```

### Step 8. Print the mean, min and max values for spirit consumption.
#### This time output a DataFrame

```python
# grouped=drinks.groupby([''])
drinks.aggregate({
    'spirit_servings':['mean','min','max']
})
```
