### Step 2. Create 3 differents Series, each of length 100, as follows: 
1. The first a random number from 1 to 4 
2. The second a random number from 1 to 3
3. The third a random number from 10,000 to 30,000

```python
import numpy as np
import pandas as pd

rng=np.random.default_rng(seed=42)

series1=pd.Series(rng.integers(low=1,high=4,size=100),name='series1')
series2=pd.Series(rng.integers(low=1,high=3,size=100),name='series2')
series3=pd.Series(rng.integers(low=10e4,high=3*10e4,size=100),name='series3')




```

### Step 3. Let's create a DataFrame by joinning the Series by column

```python
df=pd.concat([series1,series2,series3],axis=1)
# df.shape[0]
```

### Step 4. Change the name of the columns to bedrs, bathrs, price_sqr_meter

```python
# df.columns=['bedrs','bathrs','price_sql_meter']
# # df.columns
df.rename(
    columns={
        0:'bedrs',
        1:'bathrs',
        2:'price_sqr_meter'
    },
    inplace=True)
df
```

### Step 5. Create a one column DataFrame with the values of the 3 Series and assign it to 'bigcolumn'

```python
newDf=pd.DataFrame(
    {
        'bigColumn':pd.Series(np.concatenate([series1,series2,series3],axis=0))
    }
)
# newDf.describe()
newDf.shape

```
