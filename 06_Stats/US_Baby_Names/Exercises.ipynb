### Step 7. Is there more male or female names in the dataset?

```python
# from collections import Counter
baby_names.Gender.value_counts().sort_values(ascending=False)
```

### Step 8. Group the dataset by name and assign to names

```python
# names=baby_names.groupby(['Name'])['Name']
# # for name,subgroup in grouped:
# #     print(subgroup)
# #     break

#You dont want to sum the Year column,so you delete it
del baby_names['Year']

#group the data
names=baby_names.groupby('Name').sum()

#print the first 5 occurences
names.head()

#print the size of the dataset
print(names.shape)

#sort it from the biggest value to the smallest one
names.sort_values('Count',ascending=False).head()

```

### Step 9. How many different names exist in the dataset?

```python
# names.unique().count()
#as we have already grouped by the same,all the names are  unique already.
len(names) 
```

### Step 10. What is the name with most occurrences?

```python
# names.value_counts().idxmax()
names.Count.idxmax()
```

### Step 11. How many different names have the least occurrences?

```python
# names.value_counts().argmin()
len(names[names.Count==names.Count.min()])
```
