### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np

%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
```

### Step 2. Create the DataFrame that should look like the one below.

```python
df=pd.DataFrame(
    data={
        'first_name':['Jason','Molly','Tina','Jake','Amy'],
        'last_name':['Miller','Jacobson','Ali','Milner','Coze'],
        'female':[0,1,1,0,1],
        'age':[42,52,36,24,73],
        'preTestScore':[4,24,31,2,3], 
        'postTestScore':[25,94,57,62,70]
    }
)
df.head()
```

### Step 3. Create a Scatterplot of preTestScore and postTestScore, with the size of each point determined by age
#### Hint: Don't forget to place the labels

```python

sns.scatterplot(x=df.preTestScore,y=df.postTestScore,size=list(df.age),hue=df.age)

plt.xlabel('Pre Test Score')
plt.ylabel('Post Test Score')

plt.tight_layout()

plt.show()
```

### Step 4. Create a Scatterplot of preTestScore and postTestScore.
### This time the size should be 4.5 times the postTestScore and the color determined by sex

```python
sns.scatterplot(x=df.preTestScore,y=df.postTestScore,size=list(4.5*df.postTestScore),hue=df.female)

plt.tight_layout()
plt.show()
```
