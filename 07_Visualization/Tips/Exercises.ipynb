### Step 1. Import the necessary libraries:

```python
import pandas as pd
import numpy as np

%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv). 

### Step 3. Assign it to a variable called tips

```python
# tips=pd.read_csv(r'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv')
tips=pd.read_csv('tips.csv')
tips.head()
```

```python
tips.describe(include='all')
```

### Step 4. Delete the Unnamed 0 column

```python
tips.drop(
    columns=['Unnamed: 0'],inplace=True
)
```

### Step 5. Plot the total_bill column histogram

```python
sns.histplot(
    tips.total_bill,kde=True
)
# plt.xlim(10,30)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Total Bill')
plt.show()
```

```python
#create histogram
ttbill=sns.displot(tips.total_bill,kde=True)

#set labels and titles
ttbill.set(xlabel='Value',ylabel='Frequency',title='Total Bill')

#take out the right and upper borders
sns.despine()
```

### Step 6. Create a scatter plot presenting the relationship between total_bill and tip

```python
sns.scatterplot(x=tips.tip,y=tips.total_bill,hue=tips.time,size=tips.sex)

plt.xlim(2,4)


plt.show()
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips)
```

### Step 7.  Create one image with the relationship of total_bill, tip and size.

### Step 8. Present the relationship between days and total_bill value

```python
tips.day.unique()

import seaborn as sns

#group tips days
grouped=tips.groupby(['day']).sum()
# grouped.index.get_value_out(0)
grouped['day']=grouped.index.get_level_values(0)
# grouped['day']
sns.barplot(x=grouped['day'],y=grouped['total_bill'],hue=grouped['day'])
# plt.show()
```

```python
sns.stripplot(x='day',y='total_bill',data=tips,jitter=True,hue='day')
```

### Step 10.  Create a box plot presenting the total_bill per day differetiation the time (Dinner or Lunch)

```python
sns.boxplot(x='day',y='total_bill',data=tips,hue='time')
```

### Step 11. Create two histograms of the tip value based for Dinner and Lunch. They must be side by side.

```python
sns.set(style='ticks')
g=sns.FacetGrid(data=tips,col='time')
g.map(plt.hist,'tip')
```

### Step 12. Create two scatterplots graphs, one for Male and another for Female, presenting the total_bill value and tip relationship, differing by smoker or no smoker
### They must be side by side.

```python
#defining required libraries

%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns

# #categorical feature:sex,classes:Female,Male
# g=sns.FacetGrid(tips,col='total_bill',row='tip',hue='smoker')
# g.map(plt.scatter,'sex')

g=sns.FacetGrid(tips,col='sex',hue='smoker')
g.map(plt.scatter,'total_bill','tip',alpha=0.7)

g.add_legend()
```
