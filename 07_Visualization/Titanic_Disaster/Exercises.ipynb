### Step 1. Import the necessary libraries

```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns

import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Titanic_Desaster/train.csv)

### Step 3. Assign it to a variable titanic 

```python
titanic=pd.read_csv('train.csv',sep=',')
titanic.head()
```

### Step 4. Set PassengerId as the index 

```python
titanic.set_index('PassengerId')
```

### Step 5. Create a pie chart presenting the male/female proportion

```python
%matplotlib inline
import matplotlib.pyplot as plt

import seaborn as sns
# sns.set_palette()
# titanic.Sex.value_counts()

plt.pie(
    titanic.Sex.value_counts(),
    labels=['Male','Female'],
    startangle=90,
    autopct='%1.1f%%'
)

plt.title('Male and Female Proportion')
plt.legend()
```

### Step 6. Create a scatterplot with the Fare payed and the Age, differ the plot color by gender

```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
# sns.set_palette(p)
# sns.scatterplot(data=titanic,x='Fare',y='Age',hue='Sex',alpha=0.7)
sns.stripplot(data=titanic,x='Age',y='Fare',hue='Sex',jitter=True)
# plt.xlim(0,50)
# plt.ylim(0,80)

plt.xticks(np.arange(5,100,5),
           np.arange(5,100,5))
plt.tight_layout()
```

### Step 8. Create a histogram with the Fare payed

```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(
    data=titanic,x='Fare'
)
plt.xlim(0,250)
```
