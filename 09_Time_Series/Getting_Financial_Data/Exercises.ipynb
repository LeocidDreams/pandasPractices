### Step 8. Aggregate the data of volume to weekly.
Hint: Be careful to not sum data from the same week of 2015 and other years.

```python
# #sorting index (df)
# # pd.to_datetime(df.index.get_level_values(0))
# vol.index=df.index.get_level_values(0)
# # vol.head()
```

```python
# vol.sort_index(inplace=True,ascending=True)
```

```python
# #creating new dummy values with pd.daterange
# # vol.index.resample(rule='1W').mean()
# vol.resample(rule='1W').mean()
```

```python
##Solution
date=vol.index.get_level_values(0)
date=pd.DatetimeIndex(date)

vol['week']=date.isocalendar().week.values

vol['year']=date.year

vol.head()

pd.pivot_table(
    vol,values='Volume',index=['year','week'],
    columns=['stocks'],aggfunc=np.sum
)
```

### Step 9. Find all the volume traded in the year of 2015

```python
# filter_=vol.index.year==2025
# vol[filter_]
```

```python
vol_2015=vol[vol['year']==2015]

pd.pivot_table(
    vol_2015,values='Volume',index=['year'],
    columns=['stocks'],aggfunc=np.sum
)
```
